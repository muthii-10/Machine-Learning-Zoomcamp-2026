```python
# How many columns in the dataset have missing values?
count = 0
for col in df.columns:
    if df[col].isnull().any():
        count += 1
count
```

```python
# What's the maximum fuel efficiency of cars from Asia?
df[df['origin'] == 'Asia']['fuel_efficiency_mpg'].max()
```

```python
# Find the median value of the horsepower column in the dataset.
# Next, calculate the most frequent value of the same horsepower column.
# Use the fillna method to fill the missing values in the horsepower column with the most frequent value from the previous step.
# Now, calculate the median value of horsepower once again.
# Has it changed?

med1 = df['horsepower'].median()
print(f"The median value of horsepower before filling missing values is: {med1}")
mode = df['horsepower'].mode()[0]
print(f"The most frequent value of horsepower is: {mode}")
df['horsepower'].fillna(mode, inplace=True)
med2 = df['horsepower'].median()
print(f"The median value of horsepower after filling missing values is: {med2}")


if med1 == med2:
    print("The median value of horsepower has not changed.")
elif med1 < med2:
    print("The median value of horsepower has increased.")
else:
    print("The median value of horsepower has decreased.")
```

```python
# Select all the cars from Asia
# Select only columns vehicle_weight and model_year
# Select the first 7 values
# Get the underlying NumPy array. Let's call it X.
# Compute matrix-matrix multiplication between the transpose of X and X. To get the transpose, use X.T. Let's call the result XTX.
# Invert XTX.
# Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200].
# Multiply the inverse of XTX with the transpose of X, and then multiply the result by y. Call the result w.
# What's the sum of all the elements of the result?

X = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7).to_numpy()
XTX = np.dot(X.T, X)
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = np.dot(XTX_inv, np.dot(X.T, y))
result_sum = np.sum(w)
print(f"The sum of all the elements of the result is: {result_sum:.4f}")
```
